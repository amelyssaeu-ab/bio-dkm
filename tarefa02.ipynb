```python
import pandas as pd

df = pd.read_excel('dados_semana02.xlsx')

df
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

### 1. °Brix Médio por Sabor (Gráfico de Barras)

```python
plt.figure(figsize=(10, 6))
sns.barplot(x='Sabor', y='°Brix', data=df, estimator=lambda x: x.mean(), palette='viridis', hue='Sabor', legend=False)
plt.title('°Brix Médio por Sabor')
plt.xlabel('Sabor')
plt.ylabel('°Brix Médio')
plt.show()
```

### 2. Distribuição Geral de °Brix (Box Plot)

```python
plt.figure(figsize=(8, 6))
sns.boxplot(y='°Brix', data=df, color='palevioletred')
plt.title('Distribuição Geral de °Brix')
plt.ylabel('°Brix')
plt.show()
```

### 3. Distribuição de °Brix (Histograma)

```python
plt.figure(figsize=(8, 6))
sns.histplot(df['°Brix'], bins=5, kde=True, color='skyblue')
plt.title('Histograma de Distribuição de °Brix')
plt.xlabel('°Brix')
plt.ylabel('Frequência')
plt.show()
```

### 4. Distribuição de pH (Histograma)

```python
plt.figure(figsize=(8, 6))
sns.histplot(df['pH'], bins=5, kde=True, color='lightcoral')
plt.title('Histograma de Distribuição de pH')
plt.xlabel('pH')
plt.ylabel('Frequência')
plt.show()
```

###5. Distribuição de Sabores (Gráfico de Pizza)

```python
plt.figure(figsize=(9, 9))
df['Sabor'].value_counts().plot.pie(autopct='%1.1f%%', startangle=90, cmap='viridis')
plt.title('Distribuição Percentual de Sabores')
plt.ylabel('') # Remove o rótulo 'Sabor' do eixo Y que é adicionado automaticamente pelo pandas
plt.show()
```
