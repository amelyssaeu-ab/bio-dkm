##Exercício 1 — Massa corporal de mamíferos
A massa corporal de uma espécie de pequeno mamífero apresenta distribuição normal:

• μ=420 g  
• σ=55 g
	​
###1. Probabilidade de um indivíduo pesar menos de 350 g

$$ P(X<350) $$

Primeiro calculamos o \(z\): $$ z=\frac{350-420}{55}  
$$ $$ z=\frac{-70}{55} $$     
$$ z\approx -1,27 $$

Agora colocamos \(z=-1,27\) na tabela da distribuição normal: $$ P(Z<-1,27)\approx0,1020 $$

Portanto: $$ \boxed{P(X<350)\approx0,1020} $$

ou

$$ 0,1020\times100=\boxed{10,20\%} $$

**Resposta:** aproximadamente 10,2% dos mamíferos apresentam massa corporal inferior a 350 g.

###2. Probabilidade de pesar mais de 500 g
$$ P(X>500) $$

Calculamos o \(z\): $$ z=\frac{500-420}{55} $$ $$ z=\frac{80}{55} $$ $$ z\approx1,45 $$

Na tabela normal:

$$ P(Z<1,45)\approx0,9265 $$

Na probabilidade acima de 500 g: $$ P(Z>1,45)=1-0,9265 $$ $$ P(Z>1,45)=0,0735 $$

Portanto:

$$ \boxed{P(X>500)\approx0,0735} $$

ou

$$ \boxed{7,35\%} $$

**Resposta:** aproximadamente 7,35% dos indivíduos apresentam massa corporal superior a 500 g.

###3. Probabilidade de pesar entre 400 e 480 g
$$ P(400<X<480) $$

Precisamos calcular os dois valores de \(z\).

Para 400 g: $$ z_1=\frac{400-420}{55} $$ $$ z_1=\frac{-20}{55}\approx-0,36 $$

Para 480 g: $$ z_2=\frac{480-420}{55} $$ $$ z_2=\frac{60}{55}\approx1,09 $$

Então:

$$ P(400<X<480)=P(-0,36<Z<1,09) $$

Usando a tabela:

$$ P(Z<1,09)\approx0,8621 $$

e

$$ P(Z<-0,36)\approx0,3594 $$

Subtraímos:

$$ 0,8621-0,3594=0,5027 $$

Portanto:

$$ \boxed{P(400<X<480)\approx0,5027} $$

ou

$$ \boxed{50,27\%} $$

**Resposta:** aproximadamente 50,27% dos indivíduos apresentam massa corporal entre 400 e 480 g.

##Gráficos:


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Parâmetros da distribuição normal
mu = 420
sigma = 55

# Valores de x para construir a curva
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)

# Densidade da distribuição normal
y = norm.pdf(x, mu, sigma)


# =========================================================
# 1. Probabilidade de pesar menos de 350 g
# =========================================================

p1 = norm.cdf(350, mu, sigma)

plt.figure(figsize=(8, 5))

plt.plot(x, y, label='Distribuição normal')

# Área correspondente a X < 350
x1 = x[x <= 350]
y1 = norm.pdf(x1, mu, sigma)
plt.fill_between(x1, y1, alpha=0.4)

plt.axvline(350, linestyle='--', label='350 g')
plt.axvline(mu, linestyle=':', label='μ = 420 g')

plt.title(f'P(X < 350 g) = {p1*100:.2f}%')
plt.xlabel('Massa corporal (g)')
plt.ylabel('Densidade')
plt.legend()
plt.show()


# =========================================================
# 2. Probabilidade de pesar mais de 500 g
# =========================================================

p2 = 1 - norm.cdf(500, mu, sigma)

plt.figure(figsize=(8, 5))

plt.plot(x, y, label='Distribuição normal')

# Área correspondente a X > 500
x2 = x[x >= 500]
y2 = norm.pdf(x2, mu, sigma)
plt.fill_between(x2, y2, alpha=0.4)

plt.axvline(500, linestyle='--', label='500 g')
plt.axvline(mu, linestyle=':', label='μ = 420 g')

plt.title(f'P(X > 500 g) = {p2*100:.2f}%')
plt.xlabel('Massa corporal (g)')
plt.ylabel('Densidade')
plt.legend()
plt.show()


# =========================================================
# 3. Probabilidade de pesar entre 400 e 480 g
# =========================================================

p3 = norm.cdf(480, mu, sigma) - norm.cdf(400, mu, sigma)

plt.figure(figsize=(8, 5))

plt.plot(x, y, label='Distribuição normal')

# Área correspondente a 400 < X < 480
x3 = x[(x >= 400) & (x <= 480)]
y3 = norm.pdf(x3, mu, sigma)
plt.fill_between(x3, y3, alpha=0.4)

plt.axvline(400, linestyle='--', label='400 g')
plt.axvline(480, linestyle='--', label='480 g')
plt.axvline(mu, linestyle=':', label='μ = 420 g')

plt.title(f'P(400 < X < 480 g) = {p3*100:.2f}%')
plt.xlabel('Massa corporal (g)')
plt.ylabel('Densidade')
plt.legend()
plt.show()
```

Gráfico 1: P(X<350) = 10,20%

A área destacada à esquerda de 350 g representa os indivíduos que possuem massa corporal inferior a 350 g.

Gráfico 2: P(X>500) = 7,35%

A área destacada à direita de 500 g representa os indivíduos que possuem massa corporal superior a 500 g.

Gráfico 3: P(400<X<480) = 50,27%

##Exercício 2 — Percentis de tamanho corporal
O comprimento corporal desses mamíferos segue uma distribuição normal com:

• μ=150 cm    
• σ=10,5 cm

##Exercício 3 — Integração
Uma população de aves apresenta massa corporal normalmente distribuída com:

• μ=72 g    
• σ=8 g



```python
import numpy as np
from scipy.stats import norm

# Parâmetros da distribuição normal
mu = 72
sigma = 8


# 1. P(X <= 60)
p1 = norm.cdf(60, mu, sigma)

print("1. P(X <= 60):")
print(f"{p1:.4f} ou {p1*100:.2f}%")


# 2. P(X >= 85)
p2 = 1 - norm.cdf(85, mu, sigma)

print("\n2. P(X >= 85):")
print(f"{p2:.4f} ou {p2*100:.2f}%")


# 3. P(65 <= X <= 80)
p3 = norm.cdf(80, mu, sigma) - norm.cdf(65, mu, sigma)

print("\n3. P(65 <= X <= 80):")
print(f"{p3:.4f} ou {p3*100:.2f}%")


# 4. Percentis 90, 95 e 99
p90 = norm.ppf(0.90, mu, sigma)
p95 = norm.ppf(0.95, mu, sigma)
p99 = norm.ppf(0.99, mu, sigma)

print("\n4. Percentis 90, 95 e 99:")
print(f"Percentil 90: {p90:.2f} g")
print(f"Percentil 95: {p95:.2f} g")
print(f"Percentil 99: {p99:.2f} g")
```
