##Exercício 1 — Massa corporal de mamíferos
A massa corporal de uma espécie de pequeno mamífero apresenta distribuição normal:

• μ=420 g  
• σ=55 g
	​
###1. Probabilidade de um indivíduo pesar menos de 350 g

$$ P(X<350) $$

Primeiro calculamos o \(z\): $$ z=\frac{350-420}{55}  
$$ $$ z=\frac{-70}{55} $$     
$$ z\approx -1,27 $$

Agora colocamos \(z=-1,27\) na tabela da distribuição normal: $$ P(Z<-1,27)\approx0,1020 $$

Portanto: $$ \boxed{P(X<350)\approx0,1020} $$

ou

$$ 0,1020\times100=\boxed{10,20\%} $$

**Resposta:** aproximadamente 10,2% dos mamíferos apresentam massa corporal inferior a 350 g.

###2. Probabilidade de pesar mais de 500 g
$$ P(X>500) $$

Calculamos o \(z\): $$ z=\frac{500-420}{55} $$ $$ z=\frac{80}{55} $$ $$ z\approx1,45 $$

Na tabela normal:

$$ P(Z<1,45)\approx0,9265 $$

Na probabilidade acima de 500 g: $$ P(Z>1,45)=1-0,9265 $$ $$ P(Z>1,45)=0,0735 $$

Portanto:

$$ \boxed{P(X>500)\approx0,0735} $$

ou

$$ \boxed{7,35\%} $$

**Resposta:** aproximadamente 7,35% dos indivíduos apresentam massa corporal superior a 500 g.

###3. Probabilidade de pesar entre 400 e 480 g
$$ P(400<X<480) $$

Precisamos calcular os dois valores de \(z\).

Para 400 g: $$ z_1=\frac{400-420}{55} $$ $$ z_1=\frac{-20}{55}\approx-0,36 $$

Para 480 g: $$ z_2=\frac{480-420}{55} $$ $$ z_2=\frac{60}{55}\approx1,09 $$

Então:

$$ P(400<X<480)=P(-0,36<Z<1,09) $$

Usando a tabela:

$$ P(Z<1,09)\approx0,8621 $$

e

$$ P(Z<-0,36)\approx0,3594 $$

Subtraímos:

$$ 0,8621-0,3594=0,5027 $$

Portanto:

$$ \boxed{P(400<X<480)\approx0,5027} $$

ou

$$ \boxed{50,27\%} $$

**Resposta:** aproximadamente 50,27% dos indivíduos apresentam massa corporal entre 400 e 480 g.

##Gráficos:


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Parâmetros da distribuição normal
mu = 420
sigma = 55

# Valores de x para construir a curva
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)

# Densidade da distribuição normal
y = norm.pdf(x, mu, sigma)


# =========================================================
# 1. Probabilidade de pesar menos de 350 g
# =========================================================

p1 = norm.cdf(350, mu, sigma)

plt.figure(figsize=(8, 5))

plt.plot(x, y, label='Distribuição normal')

# Área correspondente a X < 350
x1 = x[x <= 350]
y1 = norm.pdf(x1, mu, sigma)
plt.fill_between(x1, y1, alpha=0.4)

plt.axvline(350, linestyle='--', label='350 g')
plt.axvline(mu, linestyle=':', label='μ = 420 g')

plt.title(f'P(X < 350 g) = {p1*100:.2f}%')
plt.xlabel('Massa corporal (g)')
plt.ylabel('Densidade')
plt.legend()
plt.show()


# =========================================================
# 2. Probabilidade de pesar mais de 500 g
# =========================================================

p2 = 1 - norm.cdf(500, mu, sigma)

plt.figure(figsize=(8, 5))

plt.plot(x, y, label='Distribuição normal')

# Área correspondente a X > 500
x2 = x[x >= 500]
y2 = norm.pdf(x2, mu, sigma)
plt.fill_between(x2, y2, alpha=0.4)

plt.axvline(500, linestyle='--', label='500 g')
plt.axvline(mu, linestyle=':', label='μ = 420 g')

plt.title(f'P(X > 500 g) = {p2*100:.2f}%')
plt.xlabel('Massa corporal (g)')
plt.ylabel('Densidade')
plt.legend()
plt.show()


# =========================================================
# 3. Probabilidade de pesar entre 400 e 480 g
# =========================================================

p3 = norm.cdf(480, mu, sigma) - norm.cdf(400, mu, sigma)

plt.figure(figsize=(8, 5))

plt.plot(x, y, label='Distribuição normal')

# Área correspondente a 400 < X < 480
x3 = x[(x >= 400) & (x <= 480)]
y3 = norm.pdf(x3, mu, sigma)
plt.fill_between(x3, y3, alpha=0.4)

plt.axvline(400, linestyle='--', label='400 g')
plt.axvline(480, linestyle='--', label='480 g')
plt.axvline(mu, linestyle=':', label='μ = 420 g')

plt.title(f'P(400 < X < 480 g) = {p3*100:.2f}%')
plt.xlabel('Massa corporal (g)')
plt.ylabel('Densidade')
plt.legend()
plt.show()
```

Gráfico 1: P(X<350) = 10,20%

A área destacada à esquerda de 350 g representa os indivíduos que possuem massa corporal inferior a 350 g.

Gráfico 2: P(X>500) = 7,35%

A área destacada à direita de 500 g representa os indivíduos que possuem massa corporal superior a 500 g.

Gráfico 3: P(400<X<480) = 50,27%

##Exercício 2 — Percentis de tamanho corporal
O comprimento corporal desses mamíferos segue uma distribuição normal com:

• μ=150 cm    
• σ=10,5 cm

1. Percentil 80 (P80): 158.84 cm
2. O comprimento que separa os 10% maiores indivíduos é: 163.46 cm
3.

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

# Parâmetros
mu = 150
sigma = 10.5

# Percentis
p50 = mu
p80 = stats.norm.ppf(0.80, loc=mu, scale=sigma)
p90 = stats.norm.ppf(0.90, loc=mu, scale=sigma)

# Criando dados para o gráfico
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)
y = stats.norm.pdf(x, loc=mu, scale=sigma)

plt.figure(figsize=(10, 6))
plt.plot(x, y, label='Distribuição Normal (Tamanho Corporal)', color='blue', lw=2)

# Plotando as linhas verticais para os percentis
plt.axvline(p50, color='gray', linestyle='--', label=f'P50 (Mediana/Média) = {p50:.1f} cm')
plt.axvline(p80, color='green', linestyle='--', label=f'P80 = {p80:.2f} cm')
plt.axvline(p90, color='red', linestyle='--', label=f'P90 = {p90:.2f} cm')

# Colorindo as respectivas áreas
plt.fill_between(x, y, where=(x >= p90), color='red', alpha=0.3, label='10% Maiores')
plt.fill_between(x, y, where=((x >= p80) & (x < p90)), color='green', alpha=0.2, label='Entre P80 e P90')

plt.title('Distribuição Normal e Percentis de Tamanho Corporal')
plt.xlabel('Comprimento Corporal (cm)')
plt.ylabel('Densidade de Probabilidade')
plt.legend(loc='upper left')
plt.grid(True, alpha=0.3)
plt.show()
```

##Exercício 3 — Integração
Uma população de aves apresenta massa corporal normalmente distribuída com:

• μ=72 g    
• σ=8 g



```python
import numpy as np
from scipy.stats import norm

# Parâmetros da distribuição normal
mu = 72
sigma = 8


# 1. P(X <= 60)
p1 = norm.cdf(60, mu, sigma)

print("1. P(X <= 60):")
print(f"{p1:.4f} ou {p1*100:.2f}%")


# 2. P(X >= 85)
p2 = 1 - norm.cdf(85, mu, sigma)

print("\n2. P(X >= 85):")
print(f"{p2:.4f} ou {p2*100:.2f}%")


# 3. P(65 <= X <= 80)
p3 = norm.cdf(80, mu, sigma) - norm.cdf(65, mu, sigma)

print("\n3. P(65 <= X <= 80):")
print(f"{p3:.4f} ou {p3*100:.2f}%")


# 4. Percentis 90, 95 e 99
p90 = norm.ppf(0.90, mu, sigma)
p95 = norm.ppf(0.95, mu, sigma)
p99 = norm.ppf(0.99, mu, sigma)

print("\n4. Percentis 90, 95 e 99:")
print(f"Percentil 90: {p90:.2f} g")
print(f"Percentil 95: {p95:.2f} g")
print(f"Percentil 99: {p99:.2f} g")

# 5. Gráfico - Regiões de Probabilidade da Massa Corporal da População de Aves
print("\n5.")
print(f"")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

# Parâmetros da distribuição das aves
mu_aves = 72
sigma_aves = 8

# Valores dos percentis calculados anteriormente
p90_aves = stats.norm.ppf(0.90, loc=mu_aves, scale=sigma_aves)
p95_aves = stats.norm.ppf(0.95, loc=mu_aves, scale=sigma_aves)
p99_aves = stats.norm.ppf(0.99, loc=mu_aves, scale=sigma_aves)

# Eixo X para plotar a curva
x_aves = np.linspace(mu_aves - 4 * sigma_aves, mu_aves + 4 * sigma_aves, 1000)
y_aves = stats.norm.pdf(x_aves, loc=mu_aves, scale=sigma_aves)

plt.figure(figsize=(12, 7))
plt.plot(x_aves, y_aves, label='Distribuição Normal (Aves)', color='darkgreen', lw=2)

# Linhas verticais dos percentis
plt.axvline(p90_aves, color='orange', linestyle='--', lw=1.5, label=f'P90 = {p90_aves:.2f} g')
plt.axvline(p95_aves, color='red', linestyle='--', lw=1.5, label=f'P95 = {p95_aves:.2f} g')
plt.axvline(p99_aves, color='purple', linestyle='--', lw=1.5, label=f'P99 = {p99_aves:.2f} g')

# Preenchimento das regiões de probabilidade no topo da distribuição
plt.fill_between(x_aves, y_aves, where=(x_aves >= p90_aves) & (x_aves < p95_aves), color='orange', alpha=0.3, label='Top 10% a 5%')
plt.fill_between(x_aves, y_aves, where=(x_aves >= p95_aves) & (x_aves < p99_aves), color='red', alpha=0.3, label='Top 5% a 1%')
plt.fill_between(x_aves, y_aves, where=(x_aves >= p99_aves), color='purple', alpha=0.3, label='Top 1% Extremo')

plt.title('Regiões de Probabilidade da Massa Corporal da População de Aves')
plt.xlabel('Massa Corporal (g)')
plt.ylabel('Densidade de Probabilidade')
plt.legend(loc='upper left')
plt.grid(True, alpha=0.3)
plt.show()
```

6. Interpretação Ecológica

* **P90 (82,25 g):** Este valor separa os 90% menores indivíduos dos 10% mais pesados. Na ecologia de aves, indivíduos que se encontram nesta faixa (acima de 82,25 g) frequentemente apresentam vantagens competitivas significativas, como maior eficiência térmica em climas frios e maior facilidade em defender territórios contra competidores menores.
* **P95 (85,16 g):** Apenas 5% da população ultrapassa este peso. Aves nesta categoria podem ser indivíduos em idade reprodutiva madura, com acesso prioritário a recursos alimentares de alta qualidade, ou fêmeas em período de acúmulo de reservas de gordura e nutrientes para a postura de ovos.
* **P99 (90,61 g):** Representa o extremo superior absoluto (apenas 1% da população pesa mais que 90,61 g). Embora indivíduos gigantes possam ser altamente dominantes no ambiente local, extremos de massa corporal também podem acarretar desvantagens ecológicas, tais como um maior custo metabólico diário para manutenção e menor agilidade no voo, aumentando o risco de predação.
