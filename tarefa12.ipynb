###**4 - Bernoulli versus Binomial versus Poisson?**


#####**4a) Se ele examinar uma única árvore:**
Trata-se da Distribuição de Bernoulli. O experimento possui apenas uma única tentativa com dois resultados possíveis: a árvore está infestada, representado como sucesso, ou não estar infestada, sendo fracasso.


#####**4b) Distribuição Binomial**, já que temos um número fixo e conhecido de tentativas independentes, no caso 50, onde se conta o número total de sucessos, sendo as árvores infestadas nesse caso, a partir da repetição da distribuição de Bernoulli.


#####**4c) A distribuição de Bernoulli** é usado em casos de uma única tentativa, tendo um resultado binário (sim ou não/ sucesso ou fracasso).
**A distribuição Binomial** é aplicada em tentativas fixas com variáveis conhecidas, contando a quantidade de sucessos dentro de uma amostra.

Já a **distribuição de Poisson** é usada em eventos dentro de um espaço temporal contínuo, sem um fixo máximo de tentativas.
