##1. Variável aleatória
É uma variável que representa numericamente um resultado que não sabemos antecipadamente qual será, associando um valor numérico a cada possível resultado de um experimento aleatório.

**Como exemplo:** Podemos citar um estudo feito com tartaruga-verde (*Chelonia mydas*), no qual foi analisada a presença de fibropapilomatose em indivíduos examinados pelo Projeto TAMAR. Entre 8.359 tartarugas avaliadas entre 2000 e 2005, foram registrados tumores em 15,41% dos indivíduos.

Nesse caso, o **experimento aleatório** consiste em selecionar uma tartaruga-verde e verificar se ela apresenta ou não fibropapilomatose.

Podemos definir a **variável aleatória (X) como a presença da doença**, sendo **(X=1) quando o indivíduo apresenta fibropapilomatose** e **(X=0) quando não apresenta**.

Assim, os **possíveis resultados da variável aleatória são 0 ou 1**.

##2. Variáveis aleatórias discretas e contínuas
As variáveis aleatórias podem ser discretas ou contínuas, de acordo com os valores que podem assumir.

**Variáveis aleatórias discretas:** São utilizadas quando o resultado do experimento pode ser representado por uma **quantidade que pode ser contada**.

**Como exemplos:** Podemos citar o número de desovas de *Chelonia mydas* em uma temporada. É uma variável discreta, pois as desovas podem ser contadas individualmente e o número de desovas assume valores inteiros. No artigo analisado, foram registradas 2.961 desovas na Ilha de Trindade e 55 em Fernando de Noronha na temporada de 2008/2009.

Número de indivíduos híbridos em uma amostra de *Eretmochelys imbricata*, que é uma variável discreta, pois corresponde à contagem de indivíduos que apresentam hibridização. Portanto, pode assumir apenas valores inteiros, como 0, 1, 2, 3, 4 e assim por diante. No estudo citado no artigo, foram analisados 119 indivíduos, dos quais 44% eram híbridos.

Número de ovos em uma desova de *Chelonia mydas*, que é uma variável discreta, pois os ovos podem ser contados individualmente e a quantidade de ovos assume valores inteiros. No artigo analisado, foi registrada uma média de 122 ovos por desova no Atol das Rocas e 125 ovos por desova na Ilha de Trindade.

**Variáveis aleatórias contínuas:** Estão relacionadas a **características que podem ser medidas e que podem assumir diferentes valores dentro de um intervalo**.

**Como exemplos:** Podemos citar a temperatura de incubação dos ovos de *Eretmochelys imbricata*, que é uma variável contínua, pois a temperatura pode assumir valores com diferentes casas decimais. Além disso, a temperatura de incubação está relacionada à determinação sexual das tartarugas marinhas, sendo que temperaturas mais elevadas estão associadas a uma maior proporção de fêmeas (F), enquanto temperaturas mais baixas estão associadas a uma maior proporção de machos (M).

Comprimento da carapaça, que é uma variável contínua porque corresponde a uma medida que pode assumir diversos valores dentro de um intervalo, como 115,2 cm, 115,25 cm ou 115,253 cm.

Tempo de incubação dos ovos, que é uma variável contínua, pois o período de incubação pode ser medido em diferentes unidades e apresentar valores fracionados, como 55,5 dias ou 55,75 dias.


## 3. Modelo de Bernoulli

### Explicação Conceitual
O **Modelo de Bernoulli** é um modelo de probabilidade discreta que descreve um experimento aleatório composto por uma única observação com exatamente **dois resultados possíveis e mutuamente exclusivos**: a ocorrência do evento de interesse (definido como **sucesso**) ou a sua não ocorrência (definida como **fracasso**).

### Identificação dos Parâmetros no Exemplo Escolhido
Usando a análise com a tartaruga-verde (*Chelonia mydas*) e a presença de fibropapilomatose, podemos definir:

* **Sucesso (X = 1):** A tartaruga-verde apresenta fibropapilomatose.
* **Fracasso (X = 0):** A tartaruga-verde NÃO apresenta fibropapilomatose.
* **Probabilidade de Sucesso (p):** **p = 0,1541** (ou 15,41%, de acordo com os dados do Projeto TAMAR).
* **Probabilidade de Fracasso (q):** q =  1 - p
=  1 - 0,1541
=  0,8459 (ou 84,59%).
* **Variável Aleatória (\(X\)):** Presença de fibropapilomatose no indivíduo, onde \(X \sim \text{Bernoulli}(p = 0{,}1541)\) e \(X \in \{0, 1\}\).

```python
import numpy as np
import scipy.stats as stats

# Parâmetro de probabilidade de sucesso (presença de fibropapilomatose)
p = 0.1541

# Definição da variável aleatória de Bernoulli
X_bernoulli = stats.bernoulli(p)

# Probabilidades teóricas dos dois resultados possíveis
prob_fracasso = X_bernoulli.pmf(0)  # P(X = 0)
prob_sucesso = X_bernoulli.pmf(1)  # P(X = 1)

# Média e Variância teóricas
media_bernoulli = X_bernoulli.mean()
variancia_bernoulli = X_bernoulli.var()

print('--- MODELO DE BERNOULLI ---')
print(f'P(X = 0) [Sem Doença / Fracasso]: {prob_fracasso:.4f}')
print(f'P(X = 1) [Com Doença / Sucesso]: {prob_sucesso:.4f}')
print(f'Média E[X]: {media_bernoulli:.4f}')
print(f'Variância Var(X): {variancia_bernoulli:.4f}')

# Simulação empírica de 1.000 observações individuais
np.random.seed(42)
simulacao = X_bernoulli.rvs(size=1000)

casos_doentes = np.sum(simulacao == 1)
casos_sadios = np.sum(simulacao == 0)

print('\n--- RESULTADO DA SIMULAÇÃO (1.000 INDIVÍDUOS) ---')
print(f'Tartarugas com fibropapilomatose (X = 1): {casos_doentes}')
print(f'Tartarugas saudáveis (X = 0): {casos_sadios}')
```

### Interpretação da Saída no Contexto Biológico
A variável aleatória de Bernoulli assume o valor **1** para indicar que a tartaruga-verde apresenta a lesão tumoral por fibropapilomatose e o valor **0** para indicar que o indivíduo está saudável. A probabilidade **p = 0,1541** representa a chance de selecionar aleatoriamente uma tartaruga doente nas condições epidemiológicas monitoradas pelo estudo.

## 4. Modelo Binomial

### Explicação Conceitual e Condições do Modelo
O **Modelo Binomial** descreve o **número total de sucessos** obtidos em um número fixo \(n\) de ensaios de Bernoulli independentes e identicamente distribuídos.

Para a aplicação adequada do modelo no cenário biológico, devem ser atendidas as seguintes condições:
1. **Número fixo de observações (n):** O tamanho da amostra é determinado previamente (ex.: n = 20 tartarugas).
2. **Dois resultados possíveis:** Cada indivíduo é classificado apenas como doente (sucesso) ou saudável (fracasso).
3. **Probabilidade de sucesso constante (p):** A chance de um indivíduo apresentar a doença é idêntica para todas as observações (p = 0,1541).
4. **Independência entre observações:** A presença de tumores em uma tartaruga não altera a probabilidade das demais apresentarem a doença.
5. **Variável aleatória de interesse (Y):** Representa o número total de indivíduos com fibropapilomatose na amostra.

A função de probabilidade é dada por:

$$P(Y = k) = \binom{n}{k} p^k (1-p)^{n-k}$$

```python
import matplotlib.pyplot as plt
import numpy as np
import scipy.stats as stats

# Parâmetros da amostragem
n = 20  # Tamanho da amostra de tartarugas
p = 0.1541  # Prevalência da doença

# Definição do modelo Binomial
Y_binomial = stats.binom(n, p)

# Média esperada e Variância
media_esperada = Y_binomial.mean()  # E[Y] = n * p
variancia_esperada = Y_binomial.var()  # Var(Y) = n * p * (1 - p)

print('--- MODELO BINOMIAL (n = 20, p = 0.1541) ---')
print(
    f'Número médio esperado de tartarugas doentes E[Y]: {media_esperada:.2f}'
)
print(f'Variância esperada Var(Y): {variancia_esperada:.2f}')

# Cálculos de probabilidades pontuais e acumuladas
prob_0 = Y_binomial.pmf(0)
prob_3 = Y_binomial.pmf(3)
prob_ate_3 = Y_binomial.cdf(3)  # P(Y <= 3)

print(f'\nP(Y = 0) [Nenhuma doente na amostra]: {prob_0:.4f} ({prob_0*100:.2f}%)')
print(
    f'P(Y = 3) [Exatamente 3 doentes na amostra]: {prob_3:.4f} ({prob_3*100:.2f}%)'
)
print(
    f'P(Y <= 3) [No máximo 3 doentes na amostra]: {prob_ate_3:.4f} ({prob_ate_3*100:.2f}%)'
)

# Gráfico da Distribuição de Probabilidade
k_valores = np.arange(0, n + 1)
probabilidades = Y_binomial.pmf(k_valores)

plt.figure(figsize=(9, 5))
plt.bar(k_valores, probabilidades, color='#1f77b4', edgecolor='black', alpha=0.8)
plt.title(
    'Distribuição Binomial: Nº de Tartarugas com Fibropapilomatose (n = 20, p = 0.1541)'
)
plt.xlabel('Número de Indivíduos Doentes na Amostra (Y)')
plt.ylabel('Probabilidade P(Y = k)')
plt.xticks(k_valores)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()
```

### Discussão sobre a Adequação das Suposições ao Cenário Biológico

1. **Interpretação dos Resultados:** Em um grupo amostrado de 20 tartarugas-verdes, espera-se encontrar em média 3,08 indivíduos com fibropapilomatose. A probabilidade de encontrar no máximo 3 indivíduos doentes na amostra é de 62,73%.
2. **Avaliação da Suposição de Independência:** Na prática de campo, se as 20 tartarugas forem capturadas juntas na mesma área de alimentação altamente poluída, a ocorrência de fibropapilomatose pode estar correlacionada entre elas (devido ao contágio viral ou estresse ambiental compartilhado). Isso violaria levemente a premissa de independência. Para contornar essa limitação, o protocolo de amostragem deve priorizar a aleatoriedade espacial e temporal.
3. **Estabilidade da Probabilidade (p):** Assumiu-se \(p\) constante 15,41%. Contudo, a prevalência pode oscilar conforme a classe etária das tartarugas (juvenis costumam apresentar maior prevalência do que adultos). O modelo se mantém válido desde que a amostragem seja restrita a uma mesma classe populacional.

# 5. Modelo de Poisson

## 5.1 Definição

O modelo de Poisson representa a **contagem de eventos** que ocorrem em um intervalo definido de tempo, área, volume ou unidade de observação.

Se $X$ é o número de eventos no intervalo, escrevemos $X \sim \text{Poisson}(\lambda)$ e

$$P(X = k) = \frac{e^{-\lambda}\,\lambda^{k}}{k!}, \qquad k = 0, 1, 2, \dots$$

- **Parâmetro:** $\lambda > 0$, a **taxa média de ocorrência** por unidade de observação.
- **Valores possíveis:** inteiros não negativos (0, 1, 2, ...), sem limite superior fixo.
- **Média e variância:** $E(X) = \text{Var}(X) = \lambda$.

**Condições usuais de aplicação:**
1. os eventos ocorrem de forma independente uns dos outros;
2. a taxa média é constante ao longo da unidade de observação;
3. dois eventos não ocorrem exatamente ao mesmo tempo/lugar;
4. a unidade de observação (tempo, área, etc.) está bem definida.

## 5.2 Contexto biológico e dados utilizados

Nas tartarugas marinhas, o **número de ninhos (desovas) por temporada** é adotado como índice de abundância populacional (Meylan, 1995, citado no artigo). O evento que vamos contar é **uma desova registrada**, e a unidade de observação será **uma noite** da temporada reprodutiva.

**Fonte dos dados:** ALMEIDA, A. P. et al. *Avaliação do Estado de Conservação da Tartaruga Marinha Chelonia mydas (Linnaeus, 1758) no Brasil*. Biodiversidade Brasileira, ano I, n. 1, 2011, p. 12-19 (dados originais: Banco de Dados TAMAR/SITAMAR).
**Link de acesso:** *(inserir o link do artigo)*
**Data de acesso:** 29/09/2026

| Local | Temporada | Desovas registradas |
|---|---|---|
| Ilha da Trindade (ES) | 2008/2009 | 2.961 |
| Atol das Rocas (RN) | 2007/2008 | 474 |
| Fernando de Noronha (PE) | 2008/2009 | 55 |

*No Atol das Rocas não houve monitoramento em 2008/2009; por isso o artigo traz a temporada anterior.*

**Duração da temporada:** o artigo informa que a atividade reprodutiva vai de **dezembro a maio (ou início de junho)**. Vamos assumir, como simplificação, **182 dias** (1º de dezembro a 31 de maio). Essa suposição é nossa, não um dado do artigo.

```python
# Dados extraídos do artigo (Almeida et al., 2011, p. 15)
dados = pd.DataFrame({
    "local": ["Ilha da Trindade (ES)", "Atol das Rocas (RN)", "Fernando de Noronha (PE)"],
    "temporada": ["2008/2009", "2007/2008", "2008/2009"],
    "desovas": [2961, 474, 55],
})

# Suposição do grupo: temporada de dezembro a maio = 182 noites
NOITES_TEMPORADA = 182

# Taxa média de desovas por noite (lambda) = total de desovas / número de noites
dados["lambda_noite"] = (dados["desovas"] / NOITES_TEMPORADA).round(2)
dados
```

**Interpretação:** o valor de $\lambda$ é o número **médio** de desovas por noite em cada local. Trindade tem uma taxa muito maior que Noronha, o que reflete o tamanho relativo das populações reprodutivas. Com $\lambda$ pequeno (Noronha), a maioria das noites tem 0 ou 1 desova; com $\lambda$ grande (Trindade), as contagens ficam concentradas em torno de valores altos.

## 5.3 Distribuição teórica de Poisson para cada local




```python
cores = sns.color_palette("viridis", 3)
fig, eixos = plt.subplots(1, 3, figsize=(15, 4))

for ax, (_, linha), cor in zip(eixos, dados.iterrows(), cores):
    lam = linha["desovas"] / NOITES_TEMPORADA
    k = np.arange(0, int(lam + 5 * np.sqrt(lam)) + 3)   # faixa de valores a exibir
    ax.bar(k, stats.poisson.pmf(k, lam), color=cor)
    ax.set_title(f"{linha['local']}\nλ = {lam:.2f} desovas/noite")
    ax.set_xlabel("Número de desovas em uma noite (k)")
    ax.set_ylabel("P(X = k)")

plt.tight_layout()
plt.show()
```

**O que o gráfico mostra:** a probabilidade de cada número de desovas em uma noite. Em Noronha ($\lambda \approx 0{,}30$) a distribuição é muito assimétrica, com o maior valor em 0. No Atol das Rocas ($\lambda \approx 2{,}60$) ela já tem um pico em 2 e 3. Em Trindade ($\lambda \approx 16{,}3$) ela é quase simétrica e parecida com uma curva em sino.

```python
# Probabilidades de interesse biológico (cálculo exato com scipy)
lam_nor = 55 / NOITES_TEMPORADA
lam_atol = 474 / NOITES_TEMPORADA
lam_tri = 2961 / NOITES_TEMPORADA

resultados = pd.DataFrame({
    "Pergunta": [
        "Noronha: nenhuma desova em uma noite, P(X = 0)",
        "Noronha: pelo menos 1 desova em uma noite, P(X ≥ 1)",
        "Atol das Rocas: mais de 5 desovas em uma noite, P(X > 5)",
        "Trindade: 25 ou mais desovas em uma noite, P(X ≥ 25)",
    ],
    "Probabilidade": [
        stats.poisson.pmf(0, lam_nor),
        1 - stats.poisson.pmf(0, lam_nor),
        stats.poisson.sf(5, lam_atol),          # sf(5) = P(X > 5)
        stats.poisson.sf(24, lam_tri),          # sf(24) = P(X >= 25)
    ],
})
resultados["Probabilidade"] = resultados["Probabilidade"].round(4)
resultados
```

**Interpretação:** em Noronha, a maior parte das noites da temporada (cerca de 74%) não teria nenhuma desova sob o modelo, e apenas cerca de 26% teriam ao menos uma. Em Trindade, noites com 25 ou mais desovas são pouco prováveis (poucos por cento), embora possíveis.

## 5.4 Simulação de uma temporada reprodutiva

> **Dados simulados.** O artigo traz apenas o **total** de desovas por temporada, não a contagem noite a noite. A simulação abaixo gera contagens diárias artificiais a partir de $\text{Poisson}(\lambda)$, com $\lambda$ calculado dos totais reais. Ela serve para **ilustrar** como seriam as contagens sob o modelo, não para representar observações reais.

```python
# Simulação: 182 noites para cada local, sorteando de uma Poisson(lambda)
simulados = {
    "Ilha da Trindade (ES)": rng.poisson(lam_tri, NOITES_TEMPORADA),
    "Atol das Rocas (RN)": rng.poisson(lam_atol, NOITES_TEMPORADA),
    "Fernando de Noronha (PE)": rng.poisson(lam_nor, NOITES_TEMPORADA),
}
lambdas = {"Ilha da Trindade (ES)": lam_tri, "Atol das Rocas (RN)": lam_atol, "Fernando de Noronha (PE)": lam_nor}

fig, eixos = plt.subplots(1, 3, figsize=(15, 4))
for ax, (local, valores) in zip(eixos, simulados.items()):
    sns.histplot(valores, discrete=True, stat="probability", kde=True,
                 color="skyblue", ax=ax, label="Simulado")
    k = np.arange(0, valores.max() + 1)
    ax.plot(k, stats.poisson.pmf(k, lambdas[local]), "o", color="lightcoral", label="Poisson teórica")
    ax.set_title(local)
    ax.set_xlabel("Desovas em uma noite")
    ax.set_ylabel("Proporção de noites")
    ax.legend()
plt.tight_layout()
plt.show()

# Média e variância simuladas (na Poisson, ambas devem ficar próximas de lambda)
resumo = pd.DataFrame({
    "lambda teórico": {l: round(v, 2) for l, v in lambdas.items()},
    "média simulada": {l: round(v.mean(), 2) for l, v in simulados.items()},
    "variância simulada": {l: round(v.var(ddof=1), 2) for l, v in simulados.items()},
})
resumo
```

**O que o gráfico mostra:** as barras azuis são a proporção de noites simuladas com cada número de desovas, e os pontos vermelhos são as probabilidades teóricas. Como a simulação usa apenas 182 noites, há pequenas diferenças, mas o formato geral coincide.

**Sobre a tabela:** a média e a variância simuladas ficam próximas de $\lambda$, refletindo a propriedade $E(X) = \text{Var}(X) = \lambda$.

## 5.5 Unidade de observação: mudando o intervalo

A taxa $\lambda$ **depende da unidade de observação**. Se os eventos seguem um processo de Poisson, ao passar de uma noite para uma semana basta multiplicar: $\lambda_{\text{semana}} = 7 \times \lambda_{\text{noite}}$. Mudar a unidade **muda o valor do parâmetro e a pergunta respondida**, mas não o modelo.

```python
comparacao_unidades = pd.DataFrame({
    "local": list(lambdas.keys()),
    "λ por noite": [round(v, 2) for v in lambdas.values()],
    "λ por semana (7 noites)": [round(7 * v, 2) for v in lambdas.values()],
})
# Probabilidade de nenhuma desova durante uma semana inteira
comparacao_unidades["P(0 desovas na semana)"] = [
    round(stats.poisson.pmf(0, 7 * v), 4) for v in lambdas.values()
]
comparacao_unidades
```

**Interpretação:** em Noronha, a chance de passar uma noite sem desova é alta (cerca de 74%), mas a chance de passar uma **semana inteira** sem nenhuma é pequena (cerca de 12%). Isso ocorre porque o intervalo maior acumula mais chances de ocorrer ao menos um evento.

## 5.6 Adequação das suposições do modelo

O uso de Poisson para desovas exige cautela:

- **Taxa não constante:** a desova não é uniforme ao longo da temporada. Há períodos de pico e períodos de poucas desovas, então $\lambda$ varia entre as noites.
- **Independência:** uma mesma fêmea pode desovar várias vezes na mesma temporada, então desovas não são totalmente independentes. Fêmeas costumam retornar em ciclos (intervalo de remigração mais frequente de 3 anos, segundo o artigo), o que também gera variação entre temporadas.
- **Dado agregado:** temos só o total da temporada, então não conseguimos testar diretamente a igualdade entre média e variância nos dados reais.

Quando $\lambda$ varia ao longo da temporada, a variância das contagens fica **maior** que a média (*sobredispersão*). O experimento abaixo ilustra esse efeito com dados simulados.

```python
# Simulação (dados simulados): taxa variável ao longo da temporada, com a mesma média total de Trindade
noites = np.arange(NOITES_TEMPORADA)
formato = np.sin(np.pi * noites / (NOITES_TEMPORADA - 1))     # pico no meio da temporada
lambda_variavel = lam_tri * formato / formato.mean()            # mesma média global de lam_tri

contagem_variavel = rng.poisson(lambda_variavel)
contagem_constante = simulados["Ilha da Trindade (ES)"]

print(f"Taxa constante  -> média = {contagem_constante.mean():.2f} | variância = {contagem_constante.var(ddof=1):.2f}")
print(f"Taxa variável   -> média = {contagem_variavel.mean():.2f} | variância = {contagem_variavel.var(ddof=1):.2f}")

fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(noites, contagem_constante, color="skyblue", label="λ constante")
ax.plot(noites, contagem_variavel, color="lightcoral", label="λ variável (pico no meio)")
ax.set_xlabel("Noite da temporada")
ax.set_ylabel("Desovas simuladas")
ax.set_title("Contagens simuladas com taxa constante e variável (Trindade)")
ax.legend()
plt.show()
```

**O que o gráfico mostra:** com taxa constante (azul), as contagens oscilam em torno de um nível fixo. Com taxa variável (vermelho), elas sobem e descem ao longo da temporada. As médias são parecidas, mas a **variância é bem maior** no segundo caso, indicando que a Poisson simples subestimaria a variabilidade real se a desova tiver picos marcados.

## 5.7 Conclusões do tópico 5

- O modelo de Poisson é adequado para descrever **contagens de eventos por unidade de observação**, como desovas por noite.
- O parâmetro $\lambda$ é a taxa média por unidade: cerca de 0,30 (Noronha), 2,60 (Atol das Rocas) e 16,27 (Trindade) desovas por noite, sob nossa suposição de 182 noites.
- Trocar a unidade de observação (noite para semana) multiplica $\lambda$ pelo tamanho do novo intervalo.
- Como a desova é sazonal e as fêmeas repetem desovas, as suposições de taxa constante e independência são **aproximações**. Para uma análise rigorosa seriam necessários dados diários reais.
- **Limitações dos dados:** os totais são de temporadas diferentes (Atol das Rocas é 2007/2008) e a duração da temporada foi assumida por nós.

# 6. Comparação entre os modelos de Bernoulli, Binomial e Poisson

Para manter o mesmo contexto biológico, usamos exemplos da própria *Chelonia mydas*. O artigo informa que foram encontrados tumores de **fibropapilomatose** em **15,41%** dos **8.359** indivíduos examinados pelo Projeto TAMAR entre 2000 e 2005 (dados de Baptistotte et al., 2005, citados no artigo). Usaremos essa proporção como probabilidade de "sucesso" (indivíduo com tumor) nos exemplos de Bernoulli e Binomial.

```python
comparacao = pd.DataFrame({
    "Modelo": ["Bernoulli", "Binomial", "Poisson"],
    "Fenômeno representado": [
        "Um único evento com dois resultados possíveis",
        "Número de sucessos em um número fixo de ensaios independentes",
        "Número de eventos em um intervalo de tempo, área ou volume",
    ],
    "Valores possíveis": [
        "0 ou 1",
        "0, 1, ..., n",
        "0, 1, 2, ... (sem limite superior)",
    ],
    "Parâmetros": [
        "p (probabilidade de sucesso)",
        "n (nº de ensaios) e p",
        "λ (taxa média por unidade)",
    ],
    "Média e variância": [
        "p  e  p(1 − p)",
        "np  e  np(1 − p)",
        "λ  e  λ",
    ],
    "Situação de aplicação": [
        "Uma observação classificada como sucesso ou fracasso",
        "Contagem de sucessos em n observações",
        "Contagem de ocorrências por unidade de observação",
    ],
    "Unidade de observação": [
        "Um indivíduo",
        "Uma amostra de n indivíduos",
        "Tempo, área, volume, armadilha, parcela etc.",
    ],
    "Exemplo em Biologia (C. mydas)": [
        "Uma tartaruga examinada tem ou não fibropapilomatose (p = 0,1541)",
        "Nº de tartarugas com tumores em uma amostra de 20 indivíduos",
        "Nº de desovas registradas por noite em Fernando de Noronha (λ ≈ 0,30)",
    ],
})

pd.set_option("display.max_colwidth", None)
comparacao.set_index("Modelo").T
```

**O que a tabela mostra:** cada modelo responde a uma pergunta diferente. Bernoulli descreve **uma** observação com dois desfechos. Binomial soma vários ensaios de Bernoulli e conta os sucessos, com máximo igual a $n$. Poisson conta ocorrências em um intervalo e **não tem teto** natural.

## 6.1 Comparação visual

Abaixo, as três distribuições lado a lado com os parâmetros dos exemplos. A amostra de $n = 20$ tartarugas é **hipotética** (escolhida por nós para ilustração); a proporção $p = 0{,}1541$ vem do artigo.

```python
p_tumor = 0.1541          # proporção de C. mydas com fibropapilomatose (artigo)
n_amostra = 20            # tamanho de amostra hipotético
lam_exemplo = lam_nor     # desovas por noite em Fernando de Noronha

cores = sns.color_palette("viridis", 3)
fig, eixos = plt.subplots(1, 3, figsize=(15, 4))

# Bernoulli
k = np.array([0, 1])
eixos[0].bar(k, stats.bernoulli.pmf(k, p_tumor), color=cores[0])
eixos[0].set_xticks(k)
eixos[0].set_xticklabels(["0 (sem tumor)", "1 (com tumor)"])
eixos[0].set_title(f"Bernoulli (p = {p_tumor})")

# Binomial
k = np.arange(0, n_amostra + 1)
eixos[1].bar(k, stats.binom.pmf(k, n_amostra, p_tumor), color=cores[1])
eixos[1].set_title(f"Binomial (n = {n_amostra}, p = {p_tumor})")
eixos[1].set_xlabel("Nº de tartarugas com tumor na amostra")

# Poisson
k = np.arange(0, 8)
eixos[2].bar(k, stats.poisson.pmf(k, lam_exemplo), color=cores[2])
eixos[2].set_title(f"Poisson (λ = {lam_exemplo:.2f})")
eixos[2].set_xlabel("Nº de desovas em uma noite")

for ax in eixos:
    ax.set_ylabel("Probabilidade")
plt.tight_layout()
plt.show()
```

**O que o gráfico mostra:** a Bernoulli tem apenas duas barras (0 e 1). A Binomial mostra a probabilidade de cada número de tartarugas com tumor numa amostra de 20, concentrada em torno de $np \approx 3{,}1$. A Poisson mostra o número de desovas por noite, com valores possíveis sem limite superior, embora as probabilidades altas sejam muito pequenas.

```python
# Verificando as médias e variâncias teóricas de cada modelo
verificacao = pd.DataFrame({
    "Média": [
        stats.bernoulli.mean(p_tumor),
        stats.binom.mean(n_amostra, p_tumor),
        stats.poisson.mean(lam_exemplo),
    ],
    "Variância": [
        stats.bernoulli.var(p_tumor),
        stats.binom.var(n_amostra, p_tumor),
        stats.poisson.var(lam_exemplo),
    ],
}, index=["Bernoulli", "Binomial", "Poisson"]).round(3)
verificacao
```

**Interpretação:** na Bernoulli e na Binomial a variância é **menor que a média** (pois $1 - p < 1$), enquanto na Poisson elas são **iguais**. Essa diferença ajuda a escolher o modelo ao analisar dados reais.

## 6.2 Relações entre os modelos

- A **Binomial** é a soma de $n$ variáveis de Bernoulli independentes com o mesmo $p$; com $n = 1$, ela se reduz à Bernoulli.
- A **Poisson** pode aproximar a Binomial quando $n$ é grande e $p$ é pequeno, com $\lambda = np$.

```python
# Demonstração numérica da aproximação (valores hipotéticos): n grande e p pequeno
n_grande, p_pequeno = 1000, 0.003
lam_aprox = n_grande * p_pequeno

k = np.arange(0, 11)
tabela = pd.DataFrame({
    "k": k,
    "Binomial(1000; 0,003)": stats.binom.pmf(k, n_grande, p_pequeno),
    "Poisson(λ = 3)": stats.poisson.pmf(k, lam_aprox),
}).round(4)
tabela
```

**Interpretação:** as duas colunas são praticamente iguais. Isso ocorre porque, com muitos ensaios e sucesso raro, contar sucessos se comporta como contar eventos raros num intervalo.

## 6.3 Conclusões do tópico 6

- **Bernoulli:** uma observação, dois desfechos, parâmetro $p$.
- **Binomial:** contagem de sucessos em $n$ ensaios fixos, com valores de 0 a $n$.
- **Poisson:** contagem de eventos por unidade de observação, sem limite superior, parâmetro $\lambda$.
- A escolha depende de haver ou não um **número fixo de ensaios**: se existe (n amostras), usa-se a Binomial; se o que se conta são ocorrências em um intervalo sem total definido, usa-se a Poisson.
